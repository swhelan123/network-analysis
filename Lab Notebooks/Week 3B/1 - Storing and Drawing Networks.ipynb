# Storing and Drawing Networks

In this notebook we will explore gasic NetworkX functionality for loading networks, visualising them, and then exporting the outputs. These capabilities will serve as building blocks for our subsequent notebooks.

```python
import networkx as nx
import matplotlib.pyplot as plt
%matplotlib inline
```

## Loading Networks

NetworkX can load existing networks stored in files using various formats, including:
- For simple edge lists, use `nx.read_edgelist()`
- For GEXF (Graph Exchange XML Format), use `nx.read_gexf()` - see https://gephi.org/gexf/format
- For GraphML, use `nx.read_graphml()` - see http://graphml.graphdrawing.org
- For DOT, use `nx.read_dot()` - see http://www.graphviz.org/content/dot-language

As an example, we can load a sample LinkedIn social network from a GEXF file using the code below. This is an undirected network where each node represents a LinkedIn user and an edge indicates that two users are connected on the social platform.

```python
g = nx.read_gexf("linkedin25.gexf")
```

```python
# check the network's size
print(f"Network has {g.number_of_nodes()} nodes and {g.number_of_edges()} edges")
```

Take a look at the node data for this network:

```python
for node, data in g.nodes(data=True):
    print(node, data)
```

## Drawing Networks

To visually display this network as a **network diagram**, we can use NetworkX's drawing capabilities which are appropriate for small networks. These use the Python Matplotlib package internally for rendering.

```python
# create a new matplotlib figure, specify its dimensions
plt.figure(figsize=(12, 8))
# add extra spacing
plt.margins(0.1, 0.1)
# display labels on the nodes
nx.draw_networkx(g, 
                 with_labels=True, 
                 font_size=11, 
                 node_size=800, 
                 node_color="lightblue")
# hide the axes/border
plt.axis("off")
plt.show()
```

NetworkX includes basic implementations of several **force-directed layout algorithms** to automatically position nodes in a network visualisation.

First, we will try the **Spring** layout:

```python
# create a new matplotlib figure, specify its dimensions
plt.figure(figsize=(12, 8))
# add extra spacing
plt.margins(0.1, 0.1)
# determine the position of the nodes based on the layout algorithm
pos = nx.spring_layout(g)
# draw the network as before, but specifiying the positions
nx.draw_networkx(g, pos, 
                 with_labels=True, 
                 node_size=800, 
                 node_color="lightblue")
plt.axis("off")
plt.show()
```

Next, we will try another force-directed algorithm, **Fruchterman-Reingold**:

```python
# create a new matplotlib figure, specify its dimensions
plt.figure(figsize=(12, 8))
# add extra spacing
plt.margins(0.1, 0.1)
# apply the layout algorithm to calculate node positions
pos = nx.fruchterman_reingold_layout(g)
# draw the network as before, but specifiying the positions
nx.draw_networkx(g, pos, 
                 with_labels=True, 
                 node_size=800, 
                 node_color="lightblue")
plt.axis("off")
plt.show()
```

## Saving and Exporting Networks

We can save a network diagram as an image (e.g. PNG file) using the standard Matplotlib `savefig()` function:

```python
plt.figure(figsize=(14, 10))
plt.margins(0.15, 0.15)
nx.draw_networkx(g, pos, 
                 with_labels=True, 
                 font_size=14, 
                 node_size=1000, 
                 node_color="lightblue")
plt.axis("off")
# save the diagram as a PNG
# the facecolour argument changes the background colour of the image
plt.savefig("linkedin-layout.png", facecolor="#cccccc")
plt.show()
```

The visualisation functionality of NetworkX is appropriate for small networks. However, this may not work well if we want to visualise and explore larger networks.

Instead, we can easily export a network in various file formats which can then be loaded by other visualisations tools, such as Gephi (https://gephi.org/):

```python
# write a comma-separate edge list format, with no attribute information
nx.write_edgelist(g, "linkedin-commas.edges", delimiter=",", data=False)
# write GraphML XML format
nx.write_graphml(g, "linkedin.graphml", prettyprint=True)
# write GEXF XML format
nx.write_gexf(g, "linkedin.gexf", prettyprint=True)
```

We can look at samples of the format of the different output files:

```python
# look at an excerpt of the edge list format
lines = open("linkedin-commas.edges", "r").readlines()
lines[0:10]
```

```python
# look at an excerpt of the GraphML format
lines = open("linkedin.graphml", "r").readlines()
lines[0:10]
```
