# Further NetworkX Visualisation

In this notebook, we explore more advanced NetworkX visualisation techniques to create more complex and informative network diagrams. We will cover customisation options including node styling, edge properties, and dynamic sizing based on network metrics.

```python
import networkx as nx
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
%matplotlib inline
```

## Loading Networks

For our examples, we will again use a sample LinkedIn social network loaded from a GEXF file:

```python
g = nx.read_gexf("linkedin25.gexf")
# check the network's size
print(f"Network has {g.number_of_nodes()} nodes and {g.number_of_edges()} edges")
```

Let's produce an initial simple visualisation of this network:

```python
# create a new matplotlib figure, specify its dimensions
plt.figure(figsize=(12, 8))
# display labels on the nodes
nx.draw_networkx(g)
# hide the axes/border
plt.axis("off")
plt.show()
```

## Customising Network Diagrams

We can customise various aspects of network visualisations to improve their clarity and information content. One approach is to draw subsets of nodes separately with different visual attributes (such as colour) to highlight specific patterns or groups in the network.

```python
# list of nodes to highlight
highlighted = ["Michael Ross", "Mike O'Hara", "Alison Gayle", "Claire Scott", "Bob O'Gara"]
```

```python
# create a new matplotlib figure, specify its dimensions
plt.figure(figsize=(12, 8))
# add extra spacing
plt.margins(0.1, 0.1)
# apply the layout algorithm to calculate node positions
pos = nx.spring_layout(g)
# draw the full network
nx.draw_networkx(g, pos, 
                 with_labels=True, 
                 font_size=12, 
                 node_size=800, 
                 node_color="lightblue")
# redraw the required nodes in a different colour
nx.draw_networkx_nodes(g, pos, 
                       nodelist=highlighted, 
                       node_size=800, 
                       node_color="yellow")
plt.axis("off")
plt.show()
```

We can also use node attributes to determine the colour of nodes programmatically. In this example, we will use the values for the 'gender' attribute to create a colour-coded visualisation.

First, we create a list of colours, one per node, based on each node's 'gender' attribute value:

```python
colours = []
# set up our mapping of gender -> colour
colourmap = {"F": "purple", "M": "green"}
# add a colour for each node based on gender
for node in g.nodes():
    colours.append(colourmap[g.nodes[node]["gender"]])
# show the list of colours we have created
print(colours)
```

We plot the network as before, but pass the list of colours to the `node_color` parameter:

```python
plt.figure(figsize=(12, 8))
plt.margins(0.1, 0.1)
pos = nx.spring_layout(g)
# draw the nodes with our custom colour scheme
nx.draw_networkx(g, pos, 
                 with_labels=True, 
                 font_size=11, 
                 node_size=900, 
                 node_color=colours)
plt.axis("off")
plt.show()
```

### Edge Styling and Transparency

We can customise the appearance of edges by modifying their colour, width, style, and transparency. 

```python
# change edge appearance with transparency and styling
plt.figure(figsize=(12, 8))
plt.margins(0.1, 0.1)
pos = nx.spring_layout(g)

# draw with custom edge styling
nx.draw_networkx(g, pos, 
                 with_labels=True, 
                 font_size=11,
                 node_size=800, 
                 node_color="lightblue",
                 edge_color="gray",
                 width=2,
                 alpha=0.7,
                 style="dashed")
plt.axis("off")
plt.show()
```

Another example with different edge styles:

```python
# we can also draw edges and nodes separately for more control
plt.figure(figsize=(12, 8))
plt.margins(0.1, 0.1)
pos = nx.spring_layout(g)

# draw edges first with custom properties
nx.draw_networkx_edges(g, pos, 
                       edge_color="red", 
                       width=3, 
                       alpha=0.3,
                       style="solid")

# draw nodes on top
nx.draw_networkx_nodes(g, pos, 
                       node_size=900, 
                       node_color="yellow",
                       alpha=0.8)

# add node labels
nx.draw_networkx_labels(g, pos, font_size=10, font_weight="bold")
plt.axis("off")
plt.show()
```

### Dynamic Node Sizing Based on Network Metrics

We can resize nodes based on various network centrality measures, such as degree centrality, to highlight structurally important nodes in the network. This provides an easy way to identify key players or influential nodes at a glance:

```python
# calculate degree centrality 
degree_centrality = nx.degree_centrality(g)
# choose proportional node sizes based on their degree centrality
node_sizes = [v * 3000 for v in degree_centrality.values()]

plt.figure(figsize=(12, 8))
plt.margins(0.1, 0.1)
pos = nx.spring_layout(g)
nx.draw_networkx(g, pos, 
                 with_labels=True, 
                 font_size=10,
                 node_size=node_sizes, 
                 node_color="lightcoral",
                 alpha=0.8)
plt.axis("off")
plt.show()
```

Another example, this time using betweenness centrality for node sizing:

```python
betweenness_centrality = nx.betweenness_centrality(g)
node_sizes_bet = [v * 5000 + 200 for v in betweenness_centrality.values()]  # +200 for minimum size

plt.figure(figsize=(12, 8))
plt.margins(0.1, 0.1)
pos = nx.spring_layout(g)
nx.draw_networkx(g, pos, 
                 with_labels=True, 
                 font_size=10,
                 node_size=node_sizes_bet, 
                 node_color="lightgreen",
                 alpha=0.8)
plt.axis("off")
plt.show()
```

### Legend Creation and Custom Colour Schemes

We can create more sophisticated colour schemes and add legends to make our visualisations more informative and interpretable. Legends are important when using colour to encode categorical or continuous data:

```python
import matplotlib.patches as mpatches

# create custom colormap for attributes with legend
plt.figure(figsize=(12, 8))
plt.margins(0.1, 0.1)
pos = nx.spring_layout(g)

# draw network with gender-based colouring
nx.draw_networkx(g, pos, 
                 with_labels=True, 
                 font_size=10,
                 node_size=900, 
                 node_color=colours,
                 alpha=0.8)

# add legend
purple_patch = mpatches.Patch(color="purple", label="Female")
green_patch = mpatches.Patch(color="green", label="Male")
plt.legend(handles=[purple_patch, green_patch], loc="upper right")
plt.axis("off")
plt.show()
```

### Multiple Node Shapes and Advanced Styling

We can use different node shapes to represent different categories of nodes, providing another dimension for encoding information such as node centrality or node attributes. This allows us to convey more information in a single visualisation:

In this example, we visualise the LinkedIn network with different node shapes based on the 'gender' node attribute:

```python
plt.figure(figsize=(12, 8))
plt.margins(0.1, 0.1)
pos = nx.spring_layout(g)

# separate nodes by gender for different shapes
male_nodes = [node for node in g.nodes() if g.nodes[node]["gender"] == "M"]
female_nodes = [node for node in g.nodes() if g.nodes[node]["gender"] == "F"]

# draw edges first
nx.draw_networkx_edges(g, pos, alpha=0.5, edge_color="gray")

# draw male nodes as squares
nx.draw_networkx_nodes(g, pos, nodelist=male_nodes, 
                       node_shape="s", node_color="lightblue", 
                       node_size=800, label="Male")

# draw female nodes as circles
nx.draw_networkx_nodes(g, pos, nodelist=female_nodes, 
                       node_shape="o", node_color="lightpink", 
                       node_size=800, label="Female")

# draw labels
nx.draw_networkx_labels(g, pos, font_size=10)
plt.legend(loc="upper right")
plt.axis("off")
plt.show()
```
