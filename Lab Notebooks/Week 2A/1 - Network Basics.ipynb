# Network Basics

In this notebook we will look at the basics of creating networks with *NetworkX*, a popular open source Python library for network analysis. 

See https://networkx.org for documentation. 

```python
# import the NetworkX module
import networkx as nx
%matplotlib inline
```

### Creating Undirected Networks

The most basic NetworkX data structure is a `Graph`, which represents an **undirected network**. To create an empty undirected network, we use the function `nx.Graph()`:

```python
g = nx.Graph()
```

We can then start to add nodes and edges to the network. Each node has a unique identifier - this can be any hashable Python value, such as a string or an integer. 

We can add one node at a time using the `add_node()` method:

```python
# add 3 nodes to the network, each with a different identifier
g.add_node("Sarah")
g.add_node("Mark")
g.add_node("Bob")
```

We can also add multiple nodes at the same time from a list using `add_nodes_from()`:

```python
# add 3 more nodes to the network
g.add_nodes_from(["Lisa", "Mary", "David"])
```

To access a list of the current nodes in the network use the `nodes()` method:

```python
list(g.nodes())
```

We can check how many nodes are currently in our network using the method `number_of_nodes()`:

```python
g.number_of_nodes()
```

Currently there are no edges connecting these nodes. We can easily add undirected edges by specifying the pair of nodes:

```python
# add 4 new edges to the network
g.add_edge("Mark", "Bob")
g.add_edge("Lisa", "Sarah")
g.add_edge("Colm", "Sarah")
g.add_edge("Bob", "Colm")
```

Now we can check how many edges are in our network using the method `number_of_edges()`:

```python
g.number_of_edges()
```

As with adding nodes, we can add multiple edges at once by specifying a list of pairs as tuples:

```python
pairs = [("Mary", "David"), ("Mark", "Mary"), ("Lisa", "Colm"), ("Bob", "Lisa"), ("Colm","David")]
g.add_edges_from(pairs)
```

```python
# how many edges are in the network now?
g.number_of_edges()
```

To access a list of the current edges in the network use the `edges()` method:

```python
list(g.edges())
```

If we create an edge involving a node that is not already in the network, it will automatically be added to the network.

```python
g.add_edge("Mary", "Robert")
```

We can check if a node or an edge exists in a network:

```python
"David" in g
```

```python
"Alison" in g
```

```python
("Mary", "Robert") in g.edges
```

```python
("Sarah", "Robert") in g.edges
```

We can draw a simple diagram of the network to inspect it using `nx.draw()`. We will focus on network visualisation in more detail later in the module.

```python
import matplotlib.pyplot as plt
nx.draw(g, with_labels=True, 
        node_color="red", 
        node_size=1700, 
        font_color="white", 
        font_size=12)
plt.show()
```

For any node, we can find the list of other nodes connected to it via an edge using the `neighbors()` function. Note that this returns an iterator:

```python
for node in g.neighbors("Mary"):
    print(node)
```

In certain types of networks we might have **self-loops** - cases where an edge exists between a node and itself.

```python
g.add_edge("Robert", "Robert")
```

```python
for node in g.neighbors("Robert"):
    print(node)
```

Removing nodes or edges has similar syntax to adding them. They can be removed individually or in batch. Note that attempting to remove a node or edge that does not exist will raise an exception.

```python
g.remove_node("Robert")
# look at list of nodes that remain
list(g.nodes())
```

```python
g.remove_nodes_from(["Sarah","Lisa"])
# look at list of nodes that remain
list(g.nodes())
```

```python
g.remove_edge("Mark", "Mary")
# look at list of edges that remain
list(g.edges())
```

Draw the final network:

```python
nx.draw(g, 
        with_labels=True, 
        node_color="red", 
        node_size=1700, 
        font_color="white", 
        font_size=12)
plt.show()
```

### Creating Directed Networks

A **directed network** is a set of nodes connected by edges, where the edges have a direction associated with them. In NetworkX, this type of network is implemented as a `DiGraph` object. To create an empty undirected network, we use the function `nx.DiGraph()`.

As with undirected networks, we call the `add_edge()` function. But now the order of the nodes matters - the arguments specify the *source* and *target* for the edge.

```python
g = nx.DiGraph()
# add 6 edges, specifying the source and target nodes as arguments
g.add_edge("Dublin", "Madrid")
g.add_edge("Dublin", "Rome")
g.add_edge("Madrid", "Dublin")
g.add_edge("London", "Dublin")
g.add_edge("Rome", "Dublin")
g.add_edge("Rome", "London")
```

Once we have created the directed network, we can check its based properties. For example, the number of nodes, number of edges, list of nodes, and list of edges:

```python
# check the number of nodes
g.number_of_nodes()
```

```python
# get a list of the current nodes
list(g.nodes())
```

```python
# check the number of edges
g.number_of_edges()
```

```python
# get list of edges in the network
list(g.edges())
```

Again, we can produce a quick diagram displaying the network:

```python
nx.draw(g, 
        with_labels=True, 
        node_color="green", 
        node_size=1700, 
        font_color="white", 
        font_size=12)
plt.show()
```

We can measure the **reciprocity** of this network - i.e. the fraction of reciprocated edges (edges pointing in both directions):

```python
nx.reciprocity(g)
```

In a directed network, we can access the **predecessors** of each node *x* -- the set of nodes which have an edge that ends at *x*. Again this function returns an iterator:

```python
for node in g.predecessors("Dublin"):
    print(node)
```

We can also access the **successors** of each node *x* -- the set of nodes which have an edge that starts at *x*. This gives the same output as `neighbors()` for a directed network.

```python
for node in g.successors("Dublin"):
    print(node)
```

```python
for node in g.neighbors("Dublin"):
    print(node)
```

Some network analysis algorithms only work with undirected networks. To convert a directed network to an undirected network, use the `to_undirected()` method. This creates a copy of the original network, where the edges no longer have direction. Note that the new network does not have duplicate edges.

```python
gundir = g.to_undirected()
```

```python
# how many edges are in the new undirected network?
gundir.number_of_edges()
```

Get a Python list containing the edge pairs in the new undirected network:

```python
list(gundir.edges())
```

Let's draw the undirected network. Notice there are no longer any arrows on the edges (i.e. they have no direction).

```python
nx.draw(gundir, 
        with_labels=True, 
        node_color="green", 
        node_size=1700, 
        font_color="white", 
        font_size=12)
plt.show()
```
